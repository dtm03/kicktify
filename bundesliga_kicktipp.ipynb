````python
from kicktipp import BundesligaPredictor
from IPython.display import display
````

## Training (vor jedem neuen Spieltag ausführen)

Standard: Saisons **2023–2026**, Backend **XGBoost**. Dauert beim ersten Mal länger wegen Downloads.

````python
predictor = BundesligaPredictor(
    seasons=tuple(range(2023, 2027)),
    model_backend="xgboost",
    use_dixon_coles=False,
)
history = predictor.load_history()
print(f"Spiele im Training: {len(history)}")
predictor.fit(history)
predictor.training_summary()
````

## Tipps zum Eintragen

- **Tipp** = vorhergesagtes Ergebnis
- **Erwartung (xG)** = erwartete Tore (nicht identisch mit dem Tipp)
- **Verletzte** = grobe Transfermarkt-Liste (beeinflusst Vorhersage leicht)

````python
tips = predictor.predict_next_matchday()
display(tips)

print("\n--- Kicktipp ---")
for _, row in tips.iterrows():
    print(f"{row['Heim']} – {row['Gast']}: {row['Tipp']}")
````
