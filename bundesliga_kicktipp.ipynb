````python
from kicktipp import BundesligaPredictor
from IPython.display import display
````

## Training (run before each new matchday)

Default: seasons **2023–2026**, backend **XGBoost**. The first run takes longer due to downloads.

````python
predictor = BundesligaPredictor(
    seasons=tuple(range(2023, 2027)),
    model_backend="xgboost",
    use_dixon_coles=False,
)
history = predictor.load_history()
print(f"Matches in training set: {len(history)}")
predictor.fit(history)
predictor.training_summary()
````

## Picks for Kicktipp

- **Pick** = predicted full-time score
- **Expected (xG)** = expected goals (not always the same as the pick)
- **Injuries (H:A)** = rough Transfermarkt counts (small effect on the prediction)

````python
tips = predictor.predict_next_matchday()
display(tips)

print("\n--- Kicktipp ---")
for _, row in tips.iterrows():
    print(f"{row['Home']} – {row['Away']}: {row['Pick']}")
````
